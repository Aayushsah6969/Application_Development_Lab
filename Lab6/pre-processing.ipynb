```python
"""
Telco Customer Churn Dataset Preprocessing Script
This script preprocesses the Telco Customer Churn dataset by:
- Dropping customer identifier columns
- Converting target variable to binary
- Handling missing values
- Encoding categorical features
- Scaling numerical features
"""

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Read the raw CSV file
print("Reading dataset...")
df = pd.read_csv('Telco_Cusomer_Churn.csv')
print(f"Original dataset shape: {df.shape}")

# Drop customer identifier column
print("\nDropping customer identifier column...")
if 'customerID' in df.columns:
    df = df.drop('customerID', axis=1)

# Convert target column 'Churn' to binary values (Yes → 1, No → 0)
print("Converting Churn column to binary values...")
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

# Handle TotalCharges if it's stored as object/string (common issue in this dataset)
if df['TotalCharges'].dtype == 'object':
    print("Converting TotalCharges to numeric...")
    df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Identify numerical and categorical columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
# Remove Churn from numerical columns as it's our target
if 'Churn' in numerical_cols:
    numerical_cols.remove('Churn')

categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print(f"\nNumerical columns: {numerical_cols}")
print(f"Categorical columns: {categorical_cols}")

# Handle missing values for numerical columns (fill with median)
print("\nHandling missing values in numerical columns...")
for col in numerical_cols:
    if df[col].isnull().sum() > 0:
        median_value = df[col].median()
        df[col].fillna(median_value, inplace=True)
        print(f"  Filled {col} missing values with median: {median_value}")

# Handle missing values for categorical columns (fill with mode)
print("\nHandling missing values in categorical columns...")
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)
        print(f"  Filled {col} missing values with mode: {mode_value}")

# Separate features and target before encoding
X = df.drop('Churn', axis=1)
y = df['Churn']

# One-Hot Encode categorical features
print("\nApplying One-Hot Encoding to categorical features...")
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=False)
print(f"Shape after encoding: {X_encoded.shape}")

# Scale numerical features using StandardScaler
print("\nScaling numerical features...")
scaler = StandardScaler()
X_encoded[numerical_cols] = scaler.fit_transform(X_encoded[numerical_cols])

# Combine processed features and target back into a single DataFrame
print("\nCombining features and target...")
processed_df = pd.concat([X_encoded, y], axis=1)

# Ensure all values are numeric
print("\nVerifying all columns are numeric...")
non_numeric_cols = processed_df.select_dtypes(exclude=['int64', 'float64', 'bool']).columns.tolist()
if len(non_numeric_cols) > 0:
    print(f"Warning: Non-numeric columns found: {non_numeric_cols}")
else:
    print("All columns are numeric ✓")

# Save the processed dataset
output_file = 'telco_churn_processed.csv'
processed_df.to_csv(output_file, index=False)

# Print summary
print(f"\n{'='*60}")
print("PREPROCESSING SUMMARY")
print(f"{'='*60}")
print(f"Original dataset shape: {df.shape}")
print(f"Processed dataset shape: {processed_df.shape}")
print(f"\nProcessed CSV file saved as: {output_file}")
print(f"{'='*60}")
```
